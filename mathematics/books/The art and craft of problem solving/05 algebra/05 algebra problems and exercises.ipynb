## Example 5.2.1 

If $x + y = xy = 3$ find $x^3 + y^3$ 

I have tried to plug in some values for *x* to see if they *y* that comes out also could sum to 3 but that has not been possible yet. 

```python
import numpy as np
```

```python
numbers_to_check = np.linspace(1, 3)
```

```python
for x_num in numbers_to_check:
    for y_num in numbers_to_check:
        sum_xy = x_num+y_num
    
        if sum_xy == 3:
            print(f'found sum = 3 at {x_num} + {y_num}')

        mult_xy = x_num*y_num

        if mult_xy == 3:
            print(f'found mult = 3 at {x_num} * {y_num}') 
```

The solution from the book asks us to keep a light touch and try to solve for the **penultimate step** which in this case is $x^2 + y^2$. How might we get that? We could square $x+y$ so that we have $3^2 = (x+y)^2 = x^2 + 2xy + y^2$. 

We also know that $xy = 3$ so we can further expand to have $9 = x^2 + y^2 + 2(3)$ which yields $x^2 + y^2 = 3$. We can write $x^3 + y^3 = (x+y)(x^2 - xy + y^2)$ and since we know all these quantities we can solve the equation.

$x^3 + y^3 = 3(3-3) = 0$

Justify each step, using commutativity and associativity in proving the following identities

1. (a + b) + (c + d) = (a + d) + (b + c)

The `communitative law` states that a + b = b + a which means the order in which we add the numbers does not impact the result. 

The `associative law` states that when three or more numbers are added, the sum is the same regardless of how the numbers are grouped. Symbolically it means that $(a + b) + c$ = $a + (b + c)$

1. By the law of associativity we might write the identity first as  $a + b + (c + d)$ 

2. By the same law me way rewrite to obtain $a + (b + c) + d$. 

3. Now we can use the communitative law to rearrange the objects so we have $a + d + (b + c)$.

4. Finally parenthesis are put around a + d so we get $(a + d) + (b + c)$

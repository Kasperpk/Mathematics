Justify each step, using commutativity and associativity in proving the following identities

1. (a + b) + (c + d) = (a + d) + (b + c)

The `communitative law` states that a + b = b + c which means the order in which we add the numbers does not impact the result. 

The `associative law` states that when three or more numbers are added, the sum is the same regardless of how the numbers are grouped. Symbolically it means that $(a + b) + c$ = $a + (b + c)$

That means for example that $(a + b) + (c + d)$ is equal to $(b + a) + (d + c)$ by the communitative law. By the law of associativity we might write the identity first as  $a + (b + c + d)$ Then we can regroup to obtain $a + (d + b + c)$ Then moving $d$ outside the parenthesis to get $a + d + (b +c)$. Finally we regroup a + d to get $(a + d) + (b + c)$

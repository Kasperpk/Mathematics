In this first exercise i will use laws of communitativity and associativity to practice justifying.

The `communitative law` states that a + b = b + a which means the order in which we add the numbers does not impact the result. 

The `associative law` states that when three or more numbers are added, the sum is the same regardless of how the numbers are grouped. Symbolically it means that $(a + b) + c$ = $a + (b + c)$

Prove the cancellation law for addition

If $a + b = a + c$ then $b = c$

Adding $-a$ to both sides yields $(-a) + a + b = (-a) + a + c$

By the associative law we get $(-a + a) + b = (-a + a) + c$

Which simplifies to $0 + b = 0 + c$, hence $b = c$

Prove that if $a + b = a$ then $b = 0$ 

adding $-a$ to both sides yields $b = a + (-a)$

We know that $a + (-a) = 0$ therefore $b = 0$

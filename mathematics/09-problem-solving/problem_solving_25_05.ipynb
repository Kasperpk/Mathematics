# Problem 2

 Deductive Logic & Pattern Recognition
 
 The Problem: A detective questions four suspects (A, B, C, and D) about a jewel theft. Each suspect makes exactly one true statement and one false statement.
 * A says: "B is the thief. C is innocent."
 * B says: "I didn't steal it. D stole it."
 * C says: "A is innocent. D is the thief."
 * D says: "B stole it. C is the thief."
 
 Who is the thief?
 
 The Trick: Create a matrix or test the truth values for each individual as the thief. Use elimination to see which scenario results in a strict 1-True/1-False split for every person.
